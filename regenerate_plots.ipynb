# Regenerador de Gráficos

Este notebook permite regenerar los gráficos de las simulaciones de DCA para cada familia de proteínas. 
Podés usarlo para personalizar títulos, ejes o simplemente para volver a generar los gráficos localmente.

```python
import sys
import os
import glob

# Asegurarse de que el notebook pueda importar desde la carpeta scripts y modules
sys.path.append('scripts')
from replotter import Replotter
```

## Configuración
Elegí la proteína y la carpeta de resultados.

```python
protein = 'Ubiquitin'  # Cambia esto por la familia que quieras
results_base = f'results/{protein}/simulations_of_frozen_alignments/'
output_dir = f'domains/{protein}/graphs/'

replotter = Replotter(output_dir)

# Buscar simulaciones disponibles
sim_dirs = sorted(glob.glob(os.path.join(results_base, 'frozen_alignment_*')))
print("Simulaciones encontradas:", [os.path.basename(d) for d in sim_dirs])
```

## Generar Gráficos de una Simulación
Podés elegir una simulación específica y generar sus gráficos de Hamming y Basin Analysis.

```python
if sim_dirs:
    sim_dir = sim_dirs[0] # Usar la primera por defecto
    timestamp = os.path.basename(sim_dir).replace('frozen_alignment_', '')
    
    print(f"Generando gráficos para {protein} ({timestamp})...")
    
    # Podés probar con diferentes sufijos: '', '_sampled_neff', '_sampled'
    suffix = '_sampled_neff' 
    
    replotter.plot_hamming_and_energy(sim_dir, suffix=suffix, subfolder=timestamp)
    replotter.plot_basin_analysis(sim_dir, suffix=suffix, subfolder=timestamp)
else:
    print("No se encontraron simulaciones para esta proteína.")
```

## Gráficos Térmicos
Si existen datos de los ensambles térmicos, podés generarlos así:

```python
thermal_dir = f'results/{protein}/thermal_ensembles/'
if os.path.exists(thermal_dir):
    print(f"Generando gráficos térmicos para {protein}...")
    replotter.plot_thermal(thermal_dir, subfolder='thermal')
else:
    print("No hay datos térmicos para esta proteína.")
```
